# 01 - Data download and cleaning
Part 1 of the assignment: download the raw Appliances Energy Prediction data, parse timestamps, check missing values, and resample to hourly.

```
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent / 'src'))

from appliance_energy import config, data
```

```
raw = data.download_raw_data()
raw.head()
```

```
hourly = data.clean_and_resample(raw, freq='h')
hourly.head()
```

```
hourly.isna().sum()
```
