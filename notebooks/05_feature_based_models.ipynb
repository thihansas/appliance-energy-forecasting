# 05 - Feature-based model
Parts 5 & 6: build sensor/weather/time/lag/rolling features, fit a gradient-boosting model, inspect feature importance.

```
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent / 'src'))

from appliance_energy import config, data, features, evaluation, plotting
from appliance_energy.models import feature_models
```

```
df = data.load_appliance_data()
ml_table = features.make_ml_table(df, target=config.TARGET)
feature_cols = features.get_feature_columns(ml_table, target=config.TARGET)
ml_table.shape
```

```
ml_train = ml_table.iloc[:-config.TEST_STEPS]
ml_test = ml_table.iloc[-config.TEST_STEPS:]

model = feature_models.fit_feature_model(ml_train[feature_cols], ml_train[config.TARGET], algorithm='xgboost')
pred = feature_models.forecast_feature_model(model, ml_test[feature_cols], ml_test.index)
pred.head()
```

```
importances = feature_models.get_feature_importance(model, feature_cols)
plotting.plot_feature_importance(importances)
```

```
feature_models.summarise_feature_group_importance(importances)
```
