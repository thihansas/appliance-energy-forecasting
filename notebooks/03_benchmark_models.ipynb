# 03 - Benchmark models
Part 3: mean, naive, seasonal naive (daily/weekly), and drift forecasts over a 24h horizon.

```
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent / 'src'))

from appliance_energy import config, data, evaluation, plotting
from appliance_energy.models import benchmarks
```

```
df = data.load_appliance_data()
y = df[config.TARGET]

train = y.iloc[:-config.TEST_STEPS]
test = y.iloc[-config.TEST_STEPS:]
len(train), len(test)
```

```
forecasts = benchmarks.all_benchmark_forecasts(train, horizon=len(test), index=test.index)
results_df = evaluation.evaluate_all(forecasts, y_true=test, y_train=train)
results_df
```

```
import pandas as pd
forecast_df = pd.DataFrame({'actual': test, **forecasts})
plotting.plot_forecasts(train, test, forecast_df)
```
